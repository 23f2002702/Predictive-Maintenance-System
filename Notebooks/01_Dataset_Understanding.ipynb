## Engineering Interpretation of Variables

### UDI

Unique identifier assigned to each observation. This variable has no engineering significance and will not contribute to machine failure prediction.

### Product ID

Unique identifier representing a specific machine product. Primarily used for tracking and identification purposes.

### Type

Machine quality category:

* L = Low Quality
* M = Medium Quality
* H = High Quality

Machine type may influence reliability, operational behavior, and failure patterns.

### Air Temperature [K]

Represents the ambient environmental temperature surrounding the machine.

Elevated ambient temperatures may increase thermal stress and influence machine operating conditions.

### Process Temperature [K]

Represents the temperature of the manufacturing process.

Abnormally high process temperatures may indicate excessive friction, inefficient operation, or developing mechanical issues.

### Rotational Speed [rpm]

Represents shaft rotational speed measured in revolutions per minute.

Rotational speed influences vibration characteristics, bearing loads, fatigue behavior, and mechanical wear.

### Torque [Nm]

Represents mechanical loading applied to the machine.

Higher torque values generally correspond to increased stress on machine components and may accelerate wear mechanisms.

### Tool Wear [min]

Represents accumulated tool usage time.

Greater tool wear indicates progressive degradation and potentially higher failure risk.

### Machine Failure

Target variable indicating whether machine failure occurred.

Value 0 = Normal operation

Value 1 = Failure

### Failure Modes

The following variables identify specific failure mechanisms:

* TWF = Tool Wear Failure
* HDF = Heat Dissipation Failure
* PWF = Power Failure
* OSF = Overstrain Failure
* RNF = Random Failure

These variables provide additional insight into the underlying causes of machine failure.

## **Target Variable Analysis**

The target variable, Machine Failure, exhibits significant class imbalance.

Out of 10,000 observations, **9,661** instances correspond to normal machine operation, while only **339** instances represent machine failures.

This results in a failure rate of approximately **3.39%**.

Such imbalance is commonly observed in industrial predictive maintenance applications because equipment failures are relatively rare compared to normal operating conditions.

The class imbalance indicates that model evaluation should not rely solely on accuracy. Additional metrics such as precision, recall, F1-score, and confusion matrices will be necessary to assess predictive performance effectively.
